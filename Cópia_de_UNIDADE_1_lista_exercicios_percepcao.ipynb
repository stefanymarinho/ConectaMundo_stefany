## Exercício 2 — regressão, testes e Fail-Fast

O código abaixo representa uma regra de desconto e uma pequena suíte de testes. Não reescreva a solução. Primeiro registre qual teste você acha que falhará quando a regra for alterada.

Depois altere somente a taxa do cupom DEVOPS20 de 20% para 30%, execute a suíte e observe onde o pipeline para. Em seguida, corrija a regra e execute novamente.

```python
def calcular_valor_final(valor_base, cupom=None, vip=False):
    if valor_base <= 0:
        raise ValueError('O valor deve ser positivo.')

    taxa_desconto = 0.0
    if cupom == 'DEVOPS20':
        taxa_desconto = 0.30  # ALTERE SOMENTE ESTA TAXA

    if vip:
        taxa_desconto += 0.05

    taxa_desconto = min(taxa_desconto, 0.25)
    return round(valor_base * (1.0 - taxa_desconto), 2)


testes = [
    {'nome': 'sem desconto', 'chamada': lambda: calcular_valor_final(100.0), 'esperado': 100.0},
    {'nome': 'cupom de 20%', 'chamada': lambda: calcular_valor_final(100.0, cupom='DEVOPS20'), 'esperado': 80.0},
    {'nome': 'cupom e VIP com teto', 'chamada': lambda: calcular_valor_final(100.0, cupom='DEVOPS20', vip=True), 'esperado': 75.0},
    {'nome': 'valor inválido', 'chamada': lambda: calcular_valor_final(0.0), 'esperado': 'ValueError'},
]

resultados = []
for teste in testes:
    try:
        obtido = teste['chamada']()
    except Exception as erro:
        obtido = type(erro).__name__

    passou = obtido == teste['esperado']
    resultados.append({
        'teste': teste['nome'],
        'esperado': teste['esperado'],
        'obtido': obtido,
        'status': 'PASSOU' if passou else 'FALHOU',
    })
    print(teste['nome'], '->', 'PASSOU' if passou else 'FALHOU')
    if not passou:
        print('FAIL-FAST: a execução foi interrompida neste teste.')
        break

display(pd.DataFrame(resultados))
```

Antes da Execução:

Primeiro teste que falhará: cupom de 20%.

Testes que deixarão de executar: cupom e VIP com teto e valor inválido.

Motivo esperado: Ao alterar a taxa de desconto de 0.20 para 0.30, a função retornará 100.0×(1.0−0.30)=70.0. Como o teste cupom de 20% espera o valor obtido de 80.0, haverá divergência. O pipeline será interrompido imediatamente no segundo teste devido à lógica de interrupção (Fail-Fast).

## Exercício 4 — métricas DORA e Quality Gate

As métricas abaixo representam deploys de uma equipe. Observe que velocidade e estabilidade precisam ser analisadas juntas.

```python
deploys = pd.DataFrame([
    {'id': 'DEP-01', 'lead_time_h': 4.5, 'incidente': False, 'recuperacao_min': 0},
    {'id': 'DEP-02', 'lead_time_h': 2.1, 'incidente': False, 'recuperacao_min': 0},
    {'id': 'DEP-03', 'lead_time_h': 8.0, 'incidente': True, 'recuperacao_min': 35},
    {'id': 'DEP-04', 'lead_time_h': 1.5, 'incidente': False, 'recuperacao_min': 0},
    {'id': 'DEP-05', 'lead_time_h': 3.2, 'incidente': False, 'recuperacao_min': 0},
    {'id': 'DEP-06', 'lead_time_h': 12.0, 'incidente': True, 'recuperacao_min': 50},
    {'id': 'DEP-07', 'lead_time_h': 2.8, 'incidente': False, 'recuperacao_min': 0},
    {'id': 'DEP-08', 'lead_time_h': 1.9, 'incidente': False, 'recuperacao_min': 0},
])

def calcular_metricas(df, dias_periodo=14):
    total = len(df)
    falhas = int(df['incidente'].sum())
    recuperacoes = df.loc[df['incidente'], 'recuperacao_min']

    return {
        'deploys_por_dia': round(total / dias_periodo, 2),
        'lead_time_medio_h': round(df['lead_time_h'].mean(), 2),
        'change_failure_rate_percent': round(falhas / total * 100, 2),
        'mttr_medio_min': round(recuperacoes.mean(), 2) if len(recuperacoes) else 0,
    }

metricas_iniciais = calcular_metricas(deploys)
display(pd.DataFrame([metricas_iniciais]))
```

```python
quality_gate = {
    'cobertura_percentual': 86.0,
    'vulnerabilidades_criticas': 0,
    'erros_de_lint': 0,
    'debito_tecnico_h': 0.5,
}

limites = {
    'cobertura_minima': 80.0,
    'debito_tecnico_maximo_h': 1.0,
}

def avaliar_quality_gate(metricas, limites):
    motivos = []
    if metricas['cobertura_percentual'] < limites['cobertura_minima']:
        motivos.append('cobertura abaixo do mínimo')
    if metricas['vulnerabilidades_criticas'] > 0:
        motivos.append('há vulnerabilidade crítica')
    if metricas['erros_de_lint'] > 0:
        motivos.append('há erro bloqueante de lint')
    if metricas['debito_tecnico_h'] > limites['debito_tecnico_maximo_h']:
        motivos.append('débito técnico acima do limite')

    status = 'APROVADO' if not motivos else 'BLOQUEADO'
    return status, motivos

status, motivos = avaliar_quality_gate(quality_gate, limites)
print('Status inicial do Quality Gate:', status)
print('Motivos:', motivos if motivos else 'nenhum')
```
